```python

# PROJECT 10: ROAD ACCIDENT HOTSPOT & SEVERITY ANALYSIS

#ADMISSION NO = 2026B15310237
#SECTION NO=AI/ML 05
# 1. Import Libraries
!pip -q install folium

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import folium
from google.colab import files
from scipy import stats

# 2. Upload Dataset
uploaded = files.upload()
file_name = next(iter(uploaded))

df = pd.read_csv(file_name, low_memory=False)

print("Dataset Loaded Successfully!")
print("Shape:", df.shape)
print(df.head())

# 3. Data Cleaning
df.columns = df.columns.str.strip().str.lower()

df = df.drop_duplicates()

print("\nMissing Values:")
print(df.isnull().sum())

# 4. Check Important Columns
print("\nDataset Columns:")
print(df.columns.tolist())

# 5. Date and Time Analysis
if 'date' in df.columns:
    df['date'] = pd.to_datetime(df['date'], dayfirst=True, errors='coerce')
    df['month'] = df['date'].dt.month
    df['weekday'] = df['date'].dt.day_name()

if 'time' in df.columns:
    df['hour'] = pd.to_datetime(
        df['time'], format='%H:%M', errors='coerce'
    ).dt.hour

# 6. Severity Analysis
if 'accident_severity' in df.columns:
    severity_map = {
        1: 'Fatal',
        2: 'Serious',
        3: 'Slight'
    }
    df['severity_name'] = pd.to_numeric(
        df['accident_severity'], errors='coerce'
    ).map(severity_map)

    print("\nAccident Severity:")
    print(df['severity_name'].value_counts())

    df['severity_name'].value_counts().plot(
        kind='bar', title='Accident Severity Distribution'
    )
    plt.xlabel("Severity")
    plt.ylabel("Number of Accidents")
    plt.show()

# 7. Hourly Accident Analysis
if 'hour' in df.columns:
    df['hour'].value_counts().sort_index().plot(
        kind='bar', figsize=(10, 5),
        title='Accidents by Hour'
    )
    plt.xlabel("Hour")
    plt.ylabel("Accidents")
    plt.show()

# 8. Weekday Analysis
if 'weekday' in df.columns:
    order = ['Monday', 'Tuesday', 'Wednesday',
             'Thursday', 'Friday', 'Saturday', 'Sunday']

    df['weekday'].value_counts().reindex(order).plot(
        kind='bar', title='Accidents by Weekday'
    )
    plt.show()

# 9. Monthly Accident Analysis
if 'month' in df.columns:
    df['month'].value_counts().sort_index().plot(
        kind='line', marker='o',
        title='Monthly Accident Analysis'
    )
    plt.xlabel("Month")
    plt.ylabel("Accidents")
    plt.show()

# 10. Weather Analysis
if 'weather_conditions' in df.columns:
    print("\nWeather Conditions:")
    print(df['weather_conditions'].value_counts())

    pd.crosstab(
        df['weather_conditions'],
        df['severity_name']
    ).plot(kind='bar', figsize=(12, 5))

    plt.title("Weather vs Accident Severity")
    plt.ylabel("Accidents")
    plt.show()

# 11. Lighting Conditions Analysis
if 'light_conditions' in df.columns:
    pd.crosstab(
        df['light_conditions'],
        df['severity_name']
    ).plot(kind='bar', figsize=(12, 5))

    plt.title("Lighting vs Accident Severity")
    plt.ylabel("Accidents")
    plt.show()

# 12. NumPy Statistical Analysis
if 'number_of_casualties' in df.columns:
    casualties = pd.to_numeric(
        df['number_of_casualties'], errors='coerce'
    ).dropna()

    print("\nCasualty Statistics")
    print("Mean:", np.mean(casualties))
    print("Median:", np.median(casualties))
    print("Standard Deviation:", np.std(casualties))

# 13. Confidence Interval for Severe Accidents
if 'severity_name' in df.columns:
    valid = df['severity_name'].dropna()
    n = len(valid)
    severe = valid.isin(['Fatal', 'Serious']).sum()

    if n > 0:
        proportion = severe / n
        margin = 1.96 * np.sqrt(
            proportion * (1 - proportion) / n
        )

        print("\nSevere Accident Percentage:",
              proportion * 100)
        print("95% Confidence Interval:",
              (max(0, proportion-margin)*100,
               min(1, proportion+margin)*100))

# 14. Geographic Hotspot Analysis
if {'latitude', 'longitude'}.issubset(df.columns):

    geo = df.copy()

    geo['latitude'] = pd.to_numeric(
        geo['latitude'], errors='coerce'
    )
    geo['longitude'] = pd.to_numeric(
        geo['longitude'], errors='coerce'
    )

    geo = geo.dropna(subset=['latitude', 'longitude'])

    geo = geo[
        geo['latitude'].between(-90, 90) &
        geo['longitude'].between(-180, 180)
    ]

    geo['location'] = (
        geo['latitude'].round(3).astype(str) + ", " +
        geo['longitude'].round(3).astype(str)
    )

    hotspots = geo.groupby('location').size().sort_values(
        ascending=False
    )

    print("\nTop 10 Accident Hotspots:")
    print(hotspots.head(10))

    if not geo.empty:
        accident_map = folium.Map(
            location=[
                geo['latitude'].mean(),
                geo['longitude'].mean()
            ],
            zoom_start=6
        )

        sample = geo.sample(min(3000, len(geo)), random_state=42)

        from folium.plugins import HeatMap

        HeatMap(
            sample[['latitude', 'longitude']].values.tolist()
        ).add_to(accident_map)

        display(accident_map)
        accident_map.save("accident_hotspot_map.html")

# 15. Missing Data Analysis
missing = df.isnull().mean() * 100

print("\nMissing Data Percentage:")
print(missing.sort_values(ascending=False))

# 16. Sensitivity Analysis
if 'severity_name' in df.columns:
    original_rate = df['severity_name'].dropna().isin(
        ['Fatal', 'Serious']
    ).mean()

    complete = df.dropna(subset=['severity_name'])

    new_rate = complete['severity_name'].isin(
        ['Fatal', 'Serious']
    ).mean()

    print("Original Severe Rate:", original_rate)
    print("After Cleaning:", new_rate)

# 17. Save Cleaned Dataset
df.to_csv("cleaned_road_accident_data.csv", index=False)

print("\nPROJECT ANALYSIS COMPLETED!")

files.download("cleaned_road_accident_data.csv")
```
